## numpy array and basics
learn about numpy

```python
import numpy as np
```

### creating array form list 

```python
arr_1d = np.array([1,2,3,4])
print("1D array :", arr_1d)

arr_2d = np.array([[1,2,3],[ 4,5,6]])
print("2D array :", arr_2d)
```

### list vs numpy array

```python
py_list = [1,2,3,4]
print("python list multiplication : ", py_list *2)

np_array= np.array([1,2,3,4]) # element wise multiplication 
print("numpy array multiplication : ", np_array *2)

import time 
start = time.time()
py_list = [i*2 for i in range(1000000)]
print("\n List operation time  : ", time.time()- start)

start = time.time()
np_array = np.arange(1000000)*2
print("\n numpy operation time  : ", time.time()- start)
```

### creating array from scratch

```python
zeros = np.zeros((5,7))# 5 rows and 7 col
print("Zeros : \n" , zeros)

ones = np.ones((3,4))
print("ones array : \n" , ones)

# constant array 
full = np.full((4,4),7) # 4x4 matrix and constant value is 7
print("full array : \n" , full)

# random values in array 

random = np.random.random((3,4))
print("random array : \n" , random)

sequence = np.arange(0,11, 2) # where 2 is steps
print("sequence array : \n" , sequence)
```

### vector , matrix and tensor

```python
vector = np.array([1,2,3,4,5])
print("Vector : " , vector)

matrix = np.array([[1,2,3],
                   [5,6,7]])
print("matrix :\n " , matrix)

tensor = np.array([[[1,2], [3,4]],
                   [[5,6],[7,8]]])
print("Tensor : \n" , tensor)
```

### array properties

```python
arr = np.array([[1,2],
                [3,4]])
print("Shape : " , arr.shape)  # gives rows and colmn
print("size : " , arr.size)  # gives number of element 
print("Dimantions : " , arr.ndim)  # gives dimantion 
print("data type : " , arr.dtype)  # gives data types
print("data type : " , arr)  # gives data types
```

### Array Reshaping

```python
arr = np.arange(12)
print("Orignal array : ", arr)

reshaped = arr.reshape((3,4))
print("Reshaped array : \n " , reshaped)

flattened = reshaped.flatten()  # its return copy array
print("flattened array : \n " , flattened)

raveled = reshaped.ravel()  # its return original array
print("reveled array : \n " , raveled)

# Transposes 
transpose = reshaped.T
print("transpose array : \n " , transpose)


```
