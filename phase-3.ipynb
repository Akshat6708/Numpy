```python
import numpy as np

# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000],  # Chai Point
])

print("=== Zomato sales analysis ===")
print("\n Salse data shape :" , sales_data.shape)
print("\n sample data for first three restaurant: \n", sales_data[0:3])
print("\n sample data for first three restaurant: \n", sales_data[:,1:])

```

```python
# now find out the total seles of all the restaurants
total_sales=  sales_data[:,1:]
yearly_total_sales = np.sum(total_sales,axis=0)
print("total sales of the all restaurant : \n ",yearly_total_sales )
total_four_year_salse= np.sum(yearly_total_sales)
print("total four year sales : " , total_four_year_salse)

# minimun  sales per restaurant

min_sale = np.min(sales_data[: , 1:], axis=1)
print("minimum sales :" , min_sale)

# maximum sales per year 
max_sale =np.max(sales_data[: , 1:] , axis=0)
print("max sales per year of all the restaurant:", max_sale)

# average sales per restaurant
avg_sale = np.mean(sales_data[: , 1:] , axis=1)
print("average sales : " , avg_sale)    

# cumulative sales 
cumsum = np.cumsum(sales_data[: , 1:], axis=1)
print("cummulative sales \n",cumsum)

# now we draw a graph of a cumulative sales

plt.figure(figsize=(10,6))
plt.plot(np.mean(cumsum , axis=0))
plt.title("average cumulative sales accross the all restaurant ")
plt.xlabel("year")
plt.ylabel("sales")
plt.grid(True)
plt.show()
```
