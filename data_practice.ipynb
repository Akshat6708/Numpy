```python
import numpy as np
```

### 1. Sales Data (Retail Store Example)

Each row = a product, each column = monthly sales.

```python
sales_data = np.array([
    [101, 230, 250, 270],   # Product A
    [150, 180, 200, 210],   # Product B
    [90,  120, 140, 160],   # Product C
    [300, 320, 330, 340]    # Product D
])

```

```python
# Q.1 How many products and months are there in the dataset?
print(np.shape(sales_data))
```

```python
# Q.2 What is the total sales of all products combined?
print(np.sum(sales_data))
```

```python
# Q.3 What is the total sales for each product?
total_sales = np.sum(sales_data ,axis=1)
print(total_sales)
```

```python
# Q.4 What is the total sales for each month?
monthly_sales = np.sum(sales_data, axis=0)
print("monthly sales :",monthly_sales)
```

```python
# Q.5 What is the average sales of each product?
average_sal = np.mean(sales_data ,axis=1)

print("average sales : ", average_sal.reshape((4,1)))
```

```python
# Q.6 Which month had the highest total sales?
best_month = np.argmax(np.sum(sales_data ,axis = 0))
print("best month sales index : ", best_month)
```

```python
# Q.7 Which product had the highest overall sales?
highest_slaes_of_the_product = np.argmax(np.sum(sales_data, axis= 1))
print("highest sales product index : ",highest_slaes_of_the_product) 
```

```python
# Q.8 What were the sales of Product B in February?
# select_feb = sales_data[: , 1]
# print("sales of Product B in February :",select_feb[1])
#  or 

print("sales of Product B in February :" ,sales_data[1,1])
```

```python
# Q.9 What were the sales of all products in March?
print(sales_data[: , 2])
```

```python
# Q.10 What were the sales of Product D across all months?

print(sales_data[3 , :])
```

```python
# Q.11 What is the cumulative sales of each product month by month?
cumsum = np.cumsum(sales_data , axis=1)
print("cumulative sales of each product month by month is :\n " ,cumsum)
```

```python
# Q.12 Which products have total sales greater than 800?
total_sales_product = np.sum(sales_data , axis=1)
print(sales_data[total_sales_product>800])
```

```python
# Q.13 What percentage of total sales does each product contribute?
total= np.sum(sales_data)
product_total = np.sum(sales_data, axis = 1)
print("persentage of each product :\n", (product_total/total)*100)
```
